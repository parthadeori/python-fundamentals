## 11. Control Statements

**i. The `if` Statement**

The `if` statement is used to execute a block of code only when a condition is **True**.

It allows programs to make decisions.

```python
# Basic if statement example
age = 18

if age >= 18:
    print("You are eligible to vote.")
```

```python
# If condition that evaluates to False

temperature = 30

if temperature > 35:
    print("It is very hot today.")
```

**ii. The `if-else` Statement**

The `if-else` statement is used when we want to execute one block of code
if the condition is **True**, and another block if the condition is **False**.

It allows programs to handle two possible outcomes.

```python
# Example 1: Age check

age = 16

if age >= 18:
    print("You are eligible to vote.")
else:
    print("You are not eligible to vote.")
```

```python
# Example 2: Number check

number = 7

if number % 2 == 0:
    print("The number is even.")
else:
    print("The number is odd.")
```

**iii. The `if-elif-else` Statement**

The `if-elif-else` statement is used when we need to check
**multiple conditions**.

It allows the program to evaluate conditions one by one.

As soon as one condition becomes True, its block runs —
and the rest are skipped.

```python
# Example 1: Grade classification

marks = int(input("Enter your marks: "))

if marks >= 90:
    print("Grade A")
elif marks >= 75:
    print("Grade B")
elif marks >= 50:
    print("Grade C")
else:
    print("Fail")
```

```python
# Example 2: Number comparison

number = 0

if number > 0:
    print("Positive number")
elif number < 0:
    print("Negative number")
else:
    print("Zero")
```

**iv. Nested `if` Statements**

A nested `if` statement means placing one `if` statement
inside another `if` statement.

It is used when a second condition needs to be checked
only if the first condition is True.

```python
# Example 1: Age and ID check

age = 20
has_id = True

if age >= 18:
    if has_id:
        print("You are allowed to enter.")
```

```python
# Example 2: Positive even number check

number = 10

if number > 0:
    if number % 2 == 0:
        print("The number is positive and even.")
```

```python
# Example 3: Nested if with else

age = 17
has_id = True

if age >= 18:
    if has_id:
        print("You are allowed to enter.")
    else:
        print("You need an ID to enter.")
else:
    print("You are underage.")
```

# 12. Comparison Operators

Comparison operators are used to compare two values.

They return either `True` or `False`.

These operators are commonly used inside control statements
like `if`, `if-else`, and loops.

---

### List of Comparison Operators

| Operator | Meaning |
|----------|----------|
| `==` | Equal to |
| `!=` | Not equal to |
| `>`  | Greater than |
| `<`  | Less than |
| `>=` | Greater than or equal to |
| `<=` | Less than or equal to |

---

Comparison operators always return a Boolean value:
- `True`
- `False`


**i. The `==` Operator (Equal To)**

The `==` operator is used to check whether two values are equal.

It returns:
- `True` if both values are equal
- `False` if they are not equal

⚠️ Do not confuse `==` with `=`

- `=` → Assignment operator (used to store values)
- `==` → Comparison operator (used to compare values)

```python
# Basic equality check

print(5 == 5)     # True
print(5 == 10)    # False
```

```python
# Using variables

a = 20
b = 20

print(a == b)     # True
```

```python
# Using == inside if statement

password = "python123"

if password == "python123":
    print("Access Granted")
```

```python
# Wrong usage (this will cause an error)

# if password = "python123":
#     print("Access Granted")
```

**Key Takeaway**

- `==` compares values
- `=` assigns values
- `==` always returns True or False

**ii. The `!=` Operator (Not Equal To)**

The `!=` operator is used to check whether two values are NOT equal.

It returns:
- `True` if the values are different
- `False` if the values are equal

It is the opposite of the `==` operator.


```python
# Basic not-equal check

print(5 != 10)    # True
print(7 != 7)     # False
```

```python
# Using variables

a = 15
b = 20

print(a != b)     # True
```

```python
# Using != inside if statement

username = "admin"

if username != "guest":
    print("Access allowed.")
```

```python
# Password check example

entered_password = "abc123"

if entered_password != "python123":
    print("Incorrect password.")
```

**Key Takeaway**

- `!=` checks if values are different
- It returns True when values are not equal
- Often used in validation and condition checks

**iii. The `>` Operator (Greater Than)**

The `>` operator checks whether the value on the left
is greater than the value on the right.

It returns:
- `True` if the left value is greater
- `False` otherwise

```python
# Basic greater than comparison

print(10 > 5)     # True
print(3 > 8)      # False
```

```python
# Using variables

a = 25
b = 20

print(a > b)      # True
```

```python
# Using > inside if statement

age = 21

if age > 18:
    print("You are above 18.")
```

```python
# Score evaluation

score = 75

if score > 50:
    print("You passed the exam.")
```

**Key Takeaway**

- `>` checks if one value is larger than another
- Returns True or False
- Commonly used in age checks, score comparisons, limits, etc.

**iv. The `<` Operator (Less Than)**

The `<` operator checks whether the value on the left
is less than the value on the right.

It returns:
- `True` if the left value is smaller
- `False` otherwise

```python
# Basic less than comparison

print(3 < 8)     # True
print(10 < 5)    # False
```

```python
# Using variables

a = 12
b = 20

print(a < b)     # True
```

```python
# Using < inside if statement

temperature = 15

if temperature < 20:
    print("It is cold outside.")
```

```python
# Budget check

balance = 500
price = 750

if balance < price:
    print("Insufficient balance.")
```

**Key Takeaway**

- `<` checks if one value is smaller than another
- Returns True or False
- Frequently used in limits, thresholds, and comparisons

**v. The `>=` Operator (Greater Than or Equal To)**

The `>=` operator checks whether the value on the left
is greater than or equal to the value on the right.

It returns:
- `True` if the left value is greater
- `True` if both values are equal
- `False` otherwise

```python
# Basic greater than or equal comparison

print(10 >= 5)    # True
print(7 >= 7)     # True
print(3 >= 8)     # False
```

```python
# Using variables

marks = 50

print(marks >= 40)   # True
```

```python
# Using >= inside if statement

age = 18

if age >= 18:
    print("You are eligible to vote.")
```

```python
# Pass/Fail check

score = 40

if score >= 40:
    print("You passed the exam.")
```

**Key Takeaway**

- `>=` includes equality
- Useful when minimum limits matter
- Common in eligibility and threshold checks

**vi. The `<=` Operator (Less Than or Equal To)**

The `<=` operator checks whether the value on the left
is less than or equal to the value on the right.

It returns:
- `True` if the left value is smaller
- `True` if both values are equal
- `False` otherwise

```python
# Basic less than or equal comparison

print(5 <= 10)    # True
print(7 <= 7)     # True
print(12 <= 3)    # False
```

```python
# Using variables

age = 16

print(age <= 18)   # True
```

```python
# Using <= inside if statement

temperature = 0

if temperature <= 0:
    print("Water may freeze.")
```

```python
# Maximum limit check

items = 5

if items <= 10:
    print("Order accepted.")
```

**Key Takeaway**

- `<=` includes equality
- Useful when checking upper limits
- Often used in range validations

# 13. Logical Operators

Logical operators are used to combine multiple conditions.

They are commonly used inside `if` statements
when we need to check more than one condition.

Logical operators return:
- `True`
- `False`

---

## Types of Logical Operators

| Operator | Meaning |
|----------|----------|
| `and` | Returns True if both conditions are True |
| `or`  | Returns True if at least one condition is True |
| `not` | Reverses the result (True becomes False, False becomes True) |

**i.The `and` Operator**

The `and` operator is used to combine two conditions.

It returns `True` only when **both conditions are True**.

If even one condition is False, the result becomes False.

```python
# Basic and operator examples

print(5 > 2 and 10 > 3)     # True
print(5 > 2 and 10 < 3)     # False
print(1 > 5 and 10 > 3)     # False
```

```python
age = 20
has_id = True

print(age >= 18 and has_id)   # True
```

```python
# Entry permission example

age = 19
has_id = True

if age >= 18 and has_id:
    print("Entry allowed.")
```

```python
# Login validation

username = "admin"
password = "python123"

if username == "admin" and password == "python123":
    print("Login successful.")
```

**Key Takeaway**

- `and` combines two conditions
- Both must be True
- Commonly used in validation checks and eligibility conditions

**ii. The `or` Operator**

The `or` operator is used to combine two conditions.

It returns `True` if at least one condition is True.

It returns `False` only when both conditions are False.

```python
# Basic or operator examples

print(5 > 2 or 10 < 3)     # True
print(1 > 5 or 10 > 3)     # True
print(1 > 5 or 2 > 8)      # False
```

```python
age = 16
has_permission = True

print(age >= 18 or has_permission)   # True
```

```python
# Weekend check example

day = "Sunday"

if day == "Saturday" or day == "Sunday":
    print("It is the weekend.")
```

```python
# Discount eligibility

is_student = True
is_senior_citizen = False

if is_student or is_senior_citizen:
    print("You are eligible for a discount.")
```

**Key Takeaway**

- `or` combines conditions
- At least one must be True
- Useful when multiple paths can satisfy a condition

**iii. The `not` Operator**

The `not` operator is used to reverse a condition.

It changes:
- True → False
- False → True

It is often used when we want to check
if something is NOT happening.

```python
# Basic not operator examples

print(not True)     # False
print(not False)    # True
```

```python
print(not (5 > 2))     # False
print(not (3 > 8))     # True
```

```python
is_logged_in = False

print(not is_logged_in)   # True
```

```python
# Login check example

is_logged_in = False

if not is_logged_in:
    print("Please log in first.")
```

**Key Takeaway**

- `not` reverses the result of a condition
- Commonly used for boolean variables
- Makes code cleaner when checking negative conditions

# 14. Loops

Loops are used to execute a block of code multiple times.

Instead of writing the same code again and again,
loops allow us to automate repetition.

Loops are useful when:
- Repeating a task multiple times
- Iterating over a sequence
- Running code until a condition changes

---

## Types of Loops in Python

1. `for` loop
2. `while` loop

**i. The `for` Loop**

The `for` loop is used to repeat a block of code
for each item in a sequence.

It is commonly used when we know
how many times the loop should run.

Python `for` loops usually work with:
- `range()`
- Strings
- Lists
- Other iterable objects

```python
# Print numbers from 1 to 5

for i in range(1, 6):
    print(i)
```

```python
# Loop through each character

for letter in "Python":
    print(letter)
```

```python
# Print even numbers from 1 to 10

for number in range(1, 11):
    if number % 2 == 0:
        print(number)
```

```python
# Loop through a list

fruits = ["apple", "banana", "mango"]

for fruit in fruits:
    print(fruit)
```

**Key Takeaway**

- `for` loop repeats code for each item in a sequence
- Often used with `range()`
- Commonly used for iteration and automation

**The `range()` Function**

The `range()` function is commonly used with `for` loops.

It generates a sequence of numbers.

`range()` does not print numbers by itself —
it generates them for the loop to use.

**range(stop)**

Starts from 0  
Ends at stop - 1

```python
# range(stop)

for i in range(5):
    print(i)
```

**range(start, stop)**

Starts from start  
Ends at stop - 1

```python
# range(start, stop)

for i in range(2, 6):
    print(i)
```

**range(start, stop, step)**

Starts from start  
Ends at stop - 1  
Increases by step

```python
# range(start, stop, step)

for i in range(1, 11, 2):
    print(i)
```

```python
# Reverse counting

for i in range(10, 0, -1):
    print(i)
```

**Key Takeaway**

- `range(stop)` → starts from 0
- `range(start, stop)` → custom start
- `range(start, stop, step)` → custom step
- Stop value is always excluded

**ii. The `while` Loop**

The `while` loop repeats a block of code
as long as a condition is True.

It is commonly used when we do NOT know
exactly how many times the loop should run.

The loop stops automatically
when the condition becomes False.

```python
# Print numbers from 1 to 5

count = 1

while count <= 5:
    print(count)
    count += 1
```

```python
# Keep asking until correct password

password = ""

while password != "python123":
    password = input("Enter password: ")

print("Access granted.")
```

```python
# This will run forever (do NOT execute)

# while True:
#     print("Infinite loop")
```

**⚠ Important Rule**

In a `while` loop:
- Always update the variable inside the loop
- Otherwise, it may become an infinite loop

---

**Key Takeaway**

- `while` runs based on a condition
- Stops when condition becomes `False`
- Useful when number of repetitions is unknown

# 15. Loop Control Statements

Loop control statements are used to change
the normal flow of a loop.

They allow us to:
- Stop a loop early
- Skip an iteration
- Do nothing temporarily

---

## Types of Loop Control Statements

1. `break`
2. `continue`
3. `pass`

**i. The `break` Statement**

The `break` statement is used to immediately
exit a loop.

When `break` is executed:
- The loop stops instantly
- The program continues after the loop
- Remaining iterations are skipped

```python
# Stop the loop when i becomes 5

for i in range(1, 11):
    if i == 5:
        break
    print(i)
```

```python
# Stop while loop at 3

count = 1

while count <= 5:
    if count == 3:
        break
    print(count)
    count += 1
```

```python
# Stop when correct password is entered

while True:
    password = input("Enter password: ")

    if password == "python123":
        print("Access granted.")
        break
```

**Key Takeaway**

- `break` stops the loop completely
- Works in both `for` and `while` loops
- Often used in search and validation logic

**ii. The `continue` Statement**

The `continue` statement is used to skip
the current iteration of a loop.

When `continue` is executed:
- The remaining code in that iteration is skipped
- The loop moves to the next iteration
- The loop does NOT stop

```python
# Skip number 5

for i in range(1, 11):
    if i == 5:
        continue
    print(i)
```

```python
# Skip printing 3

count = 0

while count < 5:
    count += 1

    if count == 3:
        continue

    print(count)
```

```python
# Print only odd numbers

for number in range(1, 11):
    if number % 2 == 0:
        continue
    print(number)
```

**Key Takeaway**

- `continue` skips the current iteration
- The loop continues running
- Useful for filtering values inside loops

**iii. The `pass` Statement**

The `pass` statement does nothing.

It is used as a placeholder when:
- A statement is required syntactically
- But no code needs to be executed yet

It allows the program to run without errors.

```python
# pass inside a loop

for i in range(5):
    if i == 3:
        pass
    print(i)
```

```python
for i in range(5):
    if i == 2:
        pass  # logic will be added later
```

```python
# Empty function example

def my_function():
    pass
```

**Key Takeaway**

- `pass` does nothing
- Used as a temporary placeholder
- Prevents syntax errors in empty blocks
