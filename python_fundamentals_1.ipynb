## 2. Comments

Comments are used to explain code and are ignored by Python during execution.

```python
# This is a single-line comment

print("Comments starts with #")
```

```python
# Comments can be written above code
# to describe what the code does

print("Comments make code readable")
```

```python
print("This will run")  # This is an inline comment
```

```python
"""
This is a multi-line comment style
It is commonly used as documentation
"""

print("Python supports multi-line comments")
print("Multi-line comments starts with three single-quotes or double quotes.")
```

## 3. Variables

Variables are used to store data in memory.

```python
x = 10
y = 5

print(x)
print(y)
```

```python
name = "Python"
version = 3.11

print(name)
print(version)
```

## 4. Variable Naming Rules

Variable names must follow specific rules in Python.

```python
# Valid variable names
name = "John"
age_21 = 21
total_marks = 450

print(name)
print(age_21)
print(total_marks)
```

```python
# Using letters, numbers, and underscores
user1 = "Admin"
_user = "Guest"
score_100 = 100

print(user1)
print(_user)
print(score_100)
```

```python
# Variable names are case-sensitive
value = 10
VALue = 20

print(value)
print(VALue)
```

```python
# Invalid variable names (will cause errors)
# 1name = "Error"
# total-marks = 90
# class = "Python"
```

## 7. The `type()` Function

The `type()` function is used to check the data type of a value or variable.

```python
x = 10
print(type(x)) # class 'int' means integer data type
```

```python
y = 3.14
print(type(y)) # class 'float' means float data type
```

```python
name = "Python"
print(type(name)) # class 'str' means string data type
```

```python
is_active = True
print(type(is_active)) # class 'bool' means boolean data type
```

```python
# Using type() with input
value = input("Enter something: ") # input function always return string data type
print(type(value))
```

## 8. Type Conversion

Type conversion is used to change one data type into another.

```python
# Converting string to integer
x = "10"
print(type(x)) # this is class 'str'

y = int(x)

print(y)
print(type(y)) # now this is converted to class 'int'
```

```python
# Converting string to float
price = "99.5"
new_price = float(price)

print(new_price)
print(type(new_price))
```

```python
# Converting number to string
age = 21 # this is class 'int'
age_str = str(age)

print(age_str) # now this is converted to class 'str'
print(type(age_str))
```

```python
# Using type conversion with input
num1 = int(input("Enter first number: "))
num2 = int(input("Enter second number: "))

print(num1 + num2) # we need to convert 'str' to 'int' in order to perform arithmetic calculation
```

```python
2# Float conversion with input
value = float(input("Enter a decimal number: "))
print(value)
```

## 9. String Concatenation

String concatenation is used to join multiple strings together.

```python
# Using + operator
first_name = "John"
last_name = "Doe"

print(first_name + last_name)
```

```python
# Adding space between strings
print(first_name + " " + last_name)
```

```python
# Concatenating strings with variables
language = "Python"
level = "Beginner"

print(language + " " + level)
```

```python
# Concatenating string and number (with conversion)
age = 21

print("Age: " + str(age)) # we need to convert 'int' to 'str' in order to concatenate with strings.
```

```python
# Using input with concatenation
name = input("Enter your name: ")

print("Welcome " + name)
```

## 10. f-Strings

f-strings are used to format and combine strings with variables easily.

```python
name = "John"
age = 21
is_learning = True

# Syntax: f"Enter anything here {float/int/str/boolean}""

print(f"Name: {name}")
print(f"Age: {age}")
print(f"Learning: {is_learning}")
```
